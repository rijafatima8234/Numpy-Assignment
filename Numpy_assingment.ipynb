```python
# 8. Vectorized Operations
import numpy as np

numbers = np.array([10, 20, 30, 40, 50])

print(numbers + 5)
print(numbers * 2)
print(numbers - 5)
print(numbers / 5)
print(numbers ** 2)
print(numbers + 10)
print(numbers * 3)
print(numbers % 3)
```

```python
# 9. Operations Between Two Arrays
import numpy as np

prices = np.array([100, 200, 300, 400])
discount = np.array([10, 20, 30, 40])

print(prices - discount)
print(prices + discount)
print(prices * discount)
print(prices / discount)
```

```python
# 10. Mathematical Functions
import numpy as np

numbers = np.array([36, 49, 64, 81])
print(np.sqrt(numbers))
print(np.power(numbers, 3))

numbers = np.array([-15, -8, 6, 12])
print(np.abs(numbers))

numbers = np.array([2.4, 5.6, 7.8, 9.3])
print(np.round(numbers))
```

```python
# 11. Aggregation Functions
import numpy as np

scores = np.array([45, 60, 75, 80, 90])

print(np.sum(scores))
print(np.mean(scores))
print(np.max(scores))
print(np.min(scores))
print(np.median(scores))
print(np.std(scores))
```

```python
# 12. Conditions in NumPy
import numpy as np

marks = np.array([35, 50, 65, 80, 95])

print(marks > 50)
print(marks >= 50)
print(marks < 80)
print(marks <= 80)
print(marks == 65)
print(marks != 65)
```

```python
# 13. Filtering Data
marks = np.array([42, 55, 68, 73, 39, 88, 95, 47])

passed = marks[marks >= 50]
print(passed)
failed = marks[marks < 50]
print(failed)
high_marks = marks[marks > 80]
print(high_marks)
average_marks = marks[(marks >= 50) & (marks <= 80)]
print(average_marks)
```
